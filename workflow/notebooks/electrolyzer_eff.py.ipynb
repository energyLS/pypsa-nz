```python
import pandas as pd
import os
```

```python
# cost_fn = '../../resources/costs_2030.csv'
cost_fn = snakemake.input.costs[0]
```

```python
cost_fn = cost_fn[:cost_fn.rfind("_") + 1]

# Generate a DataFrame with years as index and efficiencies as values
efficiency_data = []

for year in range(2020, 2051, 5):  # Loop through years 2020 to 2050 in steps of 5
    file_path = cost_fn + f"{year}.csv"
    costs = pd.read_csv(file_path)
    
    # Extract the efficiency value
    efficiency = costs.query("technology == 'electrolysis' and parameter == 'efficiency'")['value'].iloc[0]
    efficiency_data.append({"year": year, "efficiency": efficiency})

# Create a DataFrame
efficiency_df = pd.DataFrame(efficiency_data).set_index("year")
```

### Save the result

```python
efficiency_df.to_csv(snakemake.output.electrolyzer_eff)
```
